### Exercice 1

```python
def puissance(x,n):
    # cas de base
    if n == 0 :
        return 1
    # appel récursif 
    else:
        return x * puissance(x,n-1)
```

```python
# Vérification 
puissance(2,3)
```

### Exercice 2

```python
def sommeI(n):
    s=0
    for i in range(n+1) :
        s= s+i
    return s
```

```python
# Vérification
sommeI(10)
```

Dans la version récursive, le cas de base : `somme(0) = 0` et l'appel récursif : `somme(n) = n + somme(n-1)`

```python
def sommeR(n):
    if n ==0:
        return 0
    else :
        return n + sommeR(n-1)
```

```python
# Vérification
sommeR(10)
```

### Exercice 3
L'appel récursif est `pgcd(a, b) = pgcd(b, r)`  
Le cas de base : `pgcd(a, 0) = a`

```python
def pgcd(a,b):
    if b == 0 :
        return a
    else :
        r = a%b
        return pgcd(b,r)
```

```python
# Vérification
pgcd(77,132)
```

### Exercice 4
Une liste est triée si :
- Le premier élément est < au deuxième,
- La liste sans le premier élément est triée

Cas de base : une liste de 1 élément est triée

```python
def est_triee(l):
    if len(l) == 1:
        return True
    else :
        return l[0]<=l[1] and est_triee(l[1:])
```

```python
# Vérification triée
est_triee([1,2,3,4,5,6,7])
```

```python
# Vérification non triée
est_triee([1,2,3,4,6,5,7])
```

### Exercice 5

```python
def fiboI(n):
    if n==0 :
        return 1
    elif n==1:
        return 1
    a = 1
    b = 1
    c = 0
    for i in range(1,n):
        c = a + b
        a = b
        b = c
    return c
```

```python
# Vérification
fiboI(10)
```

```python
def fiboR(n):
    if n == 1 or n == 0 :
        return 1
    else:
        return fiboR(n-1) + fiboR(n-2)
```

```python
# Vérification
fiboR(10)
```

```python
from timeit import default_timer as timer
# Nombre d'éléments de la liste
nbelt=30
# Démarrage du chrono
debut_chrono = timer()
# Tri de la liste
for i in range (nbelt) :
    print (fiboI(i), end= " " )
# Fin du chrono et calcul de la durée
fin_chrono = timer()
duree = fin_chrono - debut_chrono
# Affichage du résultat
print("\n - Itératif : ",round(duree*1000),"millisecondes")
# Démarrage du chrono
debut_chrono = timer()
# Tri de la liste
for i in range (nbelt) :
    print (fiboR(i), end= " " )
# Fin du chrono et calcul de la durée
fin_chrono = timer()
duree = fin_chrono - debut_chrono
# Affichage du résultat
print("\n - Recursif : ",round(duree*1000),"millisecondes")
```

!!! info Remarque sur la performance de la récursivité
Vous remarquerez que la version récursive de la suite est beaucoup plus facile à comprendre et implémenter mais
elle n'est pas forcément la plus performante... Nous réaborderons cet aspect dans notre cours sur la
**programmation dynamique.**

### Exercice 6

Ici il y a 2 cas de base : soit le mot fait 0 lettres, soit 1 lettre, dans ces 2 cas on peut dire que c'est un palindrome.

```python
def estpalindrome(mot):
    if len(mot) == 0 :
        return True
    elif len(mot) == 1 :
        return True
    # Appel récursif : 1ere lettre = derniere lettre et reste du mot est un palindrome 
    elif mot[0] == mot[-1] and estpalindrome(mot[1:-1]) :
        return True
    else :
        return False
```

```python
# Mot avec 1 lettre à la fin
estpalindrome("kayak")
```

```python
# Mot avec 2 lettres à la fin
estpalindrome("selles")
```

```python
# Mot non palindrome
estpalindrome("maison")
```
