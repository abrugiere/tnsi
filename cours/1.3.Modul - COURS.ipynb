### Principes généraux
- Le principe de **modularité** est particulièrement important dans tout développement logiciel. Il simplifie les tests, permet de réutiliser le code et faciliter la maintenance. Ce principe peut opérer à plusieurs niveaux :
    - Découper le code en fonctions
    - Grouper les fonctions dans un fichier (module)
    - Grouper les modules en bibliothèques
- La **programmation modulaire** intervient :
    - Lors de l'écriture de portions de code pour les réutiliser plus tard ou les distribuer
    - Lors de l'utilisation, pour répondre à un besoin, de code écrit par quelqu'un d'autre

### Utilisation d'un module

!!! info Les modules
En Python il y a 3 moyens d'importer des fonctions d'un module. Regardons un exemple avec le module **`math`** et la fonction **`sqrt`** du module (**sq**uare **r**oo**t** = racine carrée) :  
_Commandes à lancer dans la console de l'interface IDLE_
!!!

#### Importer un module entier
``` python
# On déclare qu'on va utiliser les fonctions du module math
import math
# On accède aux fonctions du module en les préfixant du nom du module
print(math.sqrt(81))
```

#### Importer un module avec un alias
``` python
# On déclare qu'on va utiliser les fonctions du module math avec l'alias (raccourci) m
import math as m
# On accède aux fonctions du module en les préfixant de l'alias
print(m.sqrt(81))
```

#### Importer une fonction spécifique d'un module
``` python
# On déclare qu'on va utiliser la fonction sqrt du module math
from math import sqrt
# On accède à la fonction directement
print(sqrt(81))
```

### Création d'un module
Pour créer un module, il suffit d'écrire un programme Python contenant vos fonctions. La **docstring** du module se place au début du fichier.  
L'exemple de code ci-dessous contient deux fonctions, vous noterez que chaque fonction commence également par sa **docstring** :

``` python
""" Ce module contient des fonctions qui permettent de manipuler des nombres """

def maximum(a,b):
    """a et b sont des nombres de type int ou float la fonction renvoie le
    maximum des deux nombres"""
    if a > b:
        return a
    else:
        return b

def minimum(a,b):
    """a et b sont des nombres de type int ou float la fonction renvoie le
    minimum des deux nombres"""
    if a < b:
        return a
    else:
        return b
```

En enregistrant ce code dans un fichier de l'interface IDLE sous le nom `modNombres.py`, celui-ci sera utilisable en tant que module. Nous allons maintenant pouvoir créer un programme Python et utiliser notre module, pour cela on créera un nouveau fichier dans l'interface IDLE, pour utiliser le module il suffira de saisir le code suivant :

``` python
# Import du module modNombres
import modNombres as nb

# Lancement d'une fonction du module 
print (nb.maximum(3,5))

# Aide sur le module
help(nb)

# Aide sur une fonction du module
help(nb.minimum)

```
!!! tip Utilisation des modules
Pour pouvoir utiliser un module, le programme et le module doivent être enregistrés dans le **même répertoire**.
!!!

## Assertions

### Définition

!!! info Définition
Une **assertion** est une ligne de code qui vérifie une valeur booléenne, si cette valeur est fausse, le programme s'arrêtera et renverra une erreur.
!!!
#### Vérification des pré-conditions
Une fonction demande, pour être correcte, d'utiliser des paramètres du bon type et dans un intervalle prévu par le développeur. C'est ce qu'on appelle les **pré-conditions**. L'utilisation de mauvais paramètres peut avoir des conséquences sur l'exécution de la fonction (*exemple de StackOverflow*).  

On utilise des **assertions** pour vérifier que les valeurs passées aux paramètres (les arguments) sont du bon type.  

Dans l'exemple ci dessous, vous pouvez constater que la fonction écrite ne vérifie pas si l'entier `b` passé en paramètre est égal à $0$ et plante :

````python
def divise(a,b) :
    return a/b
````

````python
divise(1,0)
````

On peut détecter ceci en ajoutant une assertion à l'aide du mot clé `assert` :

````python
def divise(a,b) :
    assert b!=0
    return a/b
````

````python
divise(1,0)
````

Evidemment, le programme plante encore ! Mais vous pouvez constater que le message d'erreur est différent (*AssertionError*) mais peu parlant.  
Nous pouvons maintenant personnaliser le message en l'ajoutant à la suite de l'assertion sous forme de chaîne de caractères :

````python
def divise(a,b) :
    assert b!=0, 'Le second entier doit être non-nul'
    return a/b
````

````python
divise(1,0)
````

#### Tester ses fonctions

Les assertions peuvent également servir à tester votre programme, dans notre cas, on sait par exemple que `divise(6,3)` vaudra 2.  
On peut donc tester cette égalité avec une ou des assertions qu'on ajoutera à la suite de nos fonctions  :

````python
def divise(a,b) :
    assert b!=0, 'Le second entier doit être non-nul'
    return a/b

assert divise(6,3)==2, 'Erreur de codage'
````

Comme l'assertion est vraie, le programme ne renvoie pas d'erreur, mais si le programme ne renvoie pas le résultat attendu on pourra le voir avec l'assertion :

````python
def divise(a,b) :
    assert b!=0, 'Le second entier doit être non-nul'
    return a  # return erroné modifié pour lever une erreur

assert divise(6,3)==2, 'Erreur de codage'
````

Il est également possible de créer des fonctions de test pour nos fonctions, et mettre dans celle-ci tous les cas possibles :

````python
def est_pair(n):
    if n%2 == 0 :
        return True
    else :
        return False

def test_est_pair():
    assert est_pair(8) == True, 'Problème avec un nombre pair'
    assert est_pair(7) == False, 'Problème avec un nombre impair'
    print('Tests OK')
````

````python
test_est_pair()
````
