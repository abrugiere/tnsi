## Introduction

La **P**rogrammation **O**rientée **O**bjet (POO) est un paradigme de programmation élaboré dans les années 60 avec comme langage **Simula**. Mais c'est dans les années 70 avec le langage **SmallTalk** que les principes de la programmation objet sont véhiculés.  
À partir des années 1980, commence l'effervescence des langages à objets : **C++** (1983), **Objective-C** (1984), **Eiffel** (1986), **Common Lisp Object System** (1988), etc. Les années 1990 voient l'âge d'or de l'extension de la programmation par objet dans les différents secteurs du développement logiciel.

L’avantage principal de la POO, qui va pousser tous les programmeurs de l’époque à user et à abuser de ce paradigme, est la **réutilisabilité**. L’idée est d’écrire des objets informatiques autonomes, donc indépendants des programmes spécifiques qu’on écrit. De cette manière, les objets écrits pour un logiciel peuvent être directement réutilisés dans un autre programme sans modification. Cela répond donc à un défi auquel doivent faire face les informaticiens pionniers : écrire beaucoup de programmes sûrs, rapidement, et sans disposer d’une tonne de bibliothèques et de frameworks comme aujourd’hui.  

La POO est facile à apprendre mais, pour bien concevoir des programmes en POO, il faut malgré tout une bonne expérience.  

Aujourd'hui, les langages de programmation s'efforcent de tous permettre l'utilisation de différents paradigmes en même temps, Python est un langage conçu dans le paradigme de la POO (entre autres).

!!! info Définition
Le concept de base de la POO est de faire en sorte que **le code représente la vie réelle** et voir chaque entité comme un **objet**.
!!!

## Un exemple : La voiture 🚗

### Contexte

Supposons que nous voulions développer un logiciel à destination d'un concessionnaire automobile.  Pour cela nous allons devoir modéliser une voiture et stocker les informations qui composent cette voiture : 
- La marque
- Le modèle
- L'année de fabrication
- La couleur
- La puissance
- Le kilometrage

### Quelques solutions :

``` python
    # Voiture 1 = Chevrolet Corvette
    marque1 = "Chevrolet"
    modele1 = "Corvette"
    annee1 = 1973
    couleur1 = "rouge"
    puissance1 = 193

    # Voiture 2 = Renault Clio

    kilometrage3 = 12000
```

Ici on voit vite que le modèle n'est pas viable, le nombre de variables étant trop conséquent et identifier un véhicule avec les variables se terminant par le même chiffre s'avèrera vite fastidieux...  

#### Avec des tableaux :
``` python
    voiture1=["Chevrolet", "Corvette", 1973, "rouge", 193, 53000]
    
    voiture2=["Renault", "Clio", 2023, "bleu", 49, 2000]
    
    voiture3=["Mercedes", "A45 AMG", 2013, "jaune", 421, 12000]
```

💡 _Par convention une classe s'écrit toujours avec la première lettre en majuscule._





### Instance
Maintenant que nous disposons de ce nouveau type de données, il va falloir créer des objets de ce type.  

Lorsqu'en Python vous écrivez `a = 1`, Python va automatiquement attribuer un type à la variable `a` ici le type `int` pour un entier (typage dynamique). Mais tous les langages n'ont pas ce typage automatique.  

Dans le cas de la POO, pour créer une variable appelée ici  **instance** du type appelé ici **classe** on utilisera la syntaxe suivante :  
``` python
    >>> voiture1 = Voiture()
    >>> voiture2 = Voiture()
    >>> voiture3 = Voiture()
```

!!! info Définition
Une **instance** est donc un exemplaire d'un objet.
!!!


### Attributs
Nous allons maintenant vouloir que notre classe contienne les informations nécessaires correspondant aux caractéristiques de notre voiture : la marque, le modèle, l'année, la couleur, la puissance et le kilométrage.  

!!! info Définition
Les caracteristiques propres à une classe sont appelées **attributs**.
!!!



### Constructeur
Pour insérer les attributs dans une classe, une fonction interne à la classe toujours nommée `__init__` et appelée **constructeur** est utilisée, on lui passera en paramètres les valeurs des attributs. Le constructeur est toujours défini de la manière suivante :   
``` python
    class Voiture:
        """ Classe modélisant l'objet Voiture """

        def __init__(self, ma, mo, a, c, p, k):
            """ Constructeur de la classe Voiture """
            self.marque = ma
            self.modele = mo
            self.annee = a
            self.couleur = c
            self.puissance = p
            self.kilometrage = k
```

*Notez ici l'utilisation obligatoire du nom `__init__` et de la variable `self` (dont nous verrons la signification plus tard).*

Maintenant que le constructeur existe, nous allons pouvoir créer des instances en passant les attributs en paramètres lors de la création (appelée instanciation) :
``` python
    >>> voiture4 = Voiture("Dacia", "Sandero", 2017, "gris", 90, 100000)
```

!!! info Définition
La création d'un objet est appelée **instanciation** elle se traduit par l'appel du **constructeur** de la classe. Il permet de garantir que l'objet est dans un état cohérent dès sa création.
!!!

On pourra ensuite accéder aux attributs d'un objet de la manière suivante :
``` python
    >>> print(voiture4.marque)
```



!!! tip Information sur le constructeur
Il n'est pas obligatoire de passer en paramètre la valeur d'un attribut au constructeur.  
Certains attributs peuvent prendre directement une valeur par défaut dans la constructeur
!!!

On pourrait par exemple imaginer que l'objet dispose d'un attribut **immatriculation** qui au commencement aurait une valeur par défaut qui pourrait être modifiée par la suite :

``` python
    class Voiture:
        """ Classe modélisant l'objet Voiture """

        def __init__(self, ma, mo, a, c, p, k):
            """ Constructeur de la classe Voiture """
            self.marque = ma
            self.modele = mo
            self.annee = a
            self.couleur = c
            self.puissance = p
            self.kilometrage = k
            self.immatriculation = "WW-000-WW"  # Ici l'attribut est créé avec une valeur par défaut
```




### Méthodes
Supposons maintenant que nous voulions faire rouler notre voiture, ceci se traduirait par une modification de l'attribut `kilometrage` de notre véhicule (en lui ajoutant une certaine distance).  
Pour cela nous pourrions écrire une fonction qui prendrait en paramètre l'instance d'une voiture et une distance et procéderait à la modification de l'attribut concerné, par exemple :
``` python
    def faire_rouler(voiture, distance):
        voiture.kilometrage += distance

        
    >>> faire_rouler(voiture4, 500)
```

Mais cette technique ne respecte pas le paradigme objet, nous allons plutot créer une fonction **à l'intérieur** de la classe `Voiture` appelée **méthode** :
``` python
    class Voiture:
        """ Classe modélisant l'objet Voiture """

        def __init__(self, ma, mo, a, c, p, k):
            """ Constructeur de la classe Voiture """
            self.marque = ma
            self.modele = mo
            self.annee = a
            self.couleur = c
            self.puissance = p
            self.kilometrage = k
            self.immatriculation = "WW-000-WW"

        def rouler(self, distance):
            """ Methode qui permet de modifier le kilométrage """
            self.kilometrage += distance
```

*Notez ici encore l'utilisation obligatoire de la variable `self` qui représente l'instance actuelle de la classe `Voiture`.*

On pourra maintenant utiliser cette méthode de la manière suivante :
``` python
    >>> voiture4.rouler(500)
```

!!! info Définition
Une **méthode** est une procédure ou fonction propre à une classe.
!!!
