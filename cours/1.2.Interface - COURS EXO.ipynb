# Interface et implémentation

Dans les prochains cours, nous allons souvent utiliser les termes **interface** et **implémentation**. Il est important de connaitre la différence entre les deux.  


## Un exemple concret

Lorsque vous conduisez une voiture, vous apprenez à connaître son **interface**, c'est à dire les commandes qui vous permettent de **l'utiliser** : le volant, les pédales, le levier de vitesse, les commodos (phares, clignotants, essuie-glace…).  

Avec cette interface, vous êtes capables d'utiliser la voiture et vous n'êtes pas obligés de connaître comment fonctionne toute la voiture (moteur, cylindres et pistons, allumage, colonne de direction, embrayage et boîte de vitesse, électronique...). Cet aspect qui dit _"Comment c'est construit ?"_ est l'**implémentation** du véhicule, il peut changer selon le type de véhicule (voiture, camion, utilitaire, bus…) ou le type de carburant utilisé (diesel, essence, gpl, éthanol…) mais les interfaces resteront les mêmes.


## En informatique

Dans le monde informatique c'est la même chose : vous pouvez utiliser des objets ou algorithmes en connaissant leur **interface**, c'est à dire l'ensemble des actions et propriétés disponibles, sans pour autant connaître comment ces objets ont été codés et stockés en mémoire : son **implémentation**.

!!! info Définition
**L’interface** d’une structure de données ou d'une fonction est la spécification des méthodes, paramètres et sorties disponibles pour cette structure ou fonction.
!!!

- Pour un tableau de type `list` vous disposez des méthodes `.append()` pour ajouter des éléments, `.pop()` pour retirer des éléments, de la fonction `len` pour avoir le nombre d'éléments…  
Vous connaissez ces outils mais vous ne connaissez pas le code de ces outils, ceci ne vous empèche pas de les utiliser.

!!! info Définition
**L’implémentation** d’une structure de données ou d’un algorithme est sa mise en oeuvre pratique dans un langage de programmation.
!!!

- Dans notre cours sur la récursivité, nous avons écrit 2 fonctions permettant de calculer la **factorielle** d'un nombre : la version itérative et la version récursive. Ces deux fonctions sont codées différemment : elles ont donc deux implémentations différente. Cependant, elles ont la même interface : elles prennent un entier en entrée et renvoient un entier qui sera la fatorielle de ce nombre, dans les deux cas on écrira `factorielle(n)` pour les appeler et le retour sera un entier dans les deux cas.

!!! success Interfaces et implémentation : résumé

**Ce qu’il faut savoir**
- L'**interface** est l'ensemble des actions disponibles pour un algorithme ou une structure de données. Elle est nécessaire pour savoir utiliser la structure ou l'algorithme.
- L'**impémentation** est la mise en oeuvre de cette interface ou algorithme dans un langage donné. Sa connaissance n'est pas nécessaire pour utiliser une structure ou un algorithme.

**Ce qu’il faut savoir faire**
- Être capable de spécifier une structure de données par son interface.
- Être capable d’écrire plusieurs implémentations d’une même structure de données ou algorithme.
!!!
