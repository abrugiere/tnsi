#### 1. Mystère

```python
# Fonction buggée
def mystere(li):
    for i in range (len(li)):
        if li[i+1] < li[i]:
            return False
    return True
```

```python
# Vérification
mystere([1,2,3,4,5,6])
```

```python
# Vérification
mystere([1,2,3,4,5,6])
```

#### 2. Animal

```python
# Fonction buggée
def animal(prenom):
    if prenom[0] in ['A','B','D','G','H','I','K','L','O','P','V']:
        print("Tu es un lion")
    elif prenom[0] in ['C','E','F','J','M','N']:
        print("Tu es un dauphin")
```

```python
# Vérification
animal('Arthur')
```

```python
# Vérification
animal('Ryan')
```

```python
# Vérification
animal('Ryan')
```
