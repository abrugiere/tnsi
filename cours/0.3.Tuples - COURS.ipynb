# Types construits : Tuples et tableau indexés

---
En classe de première, nous avons vu les différents types simples permettant de stocker de l’information : **int**, **float**, **bool**.  

Avec ces types simples, nous pouvons définir des variables qui représenteront chacune **une** donnée avec **une** valeur. Ceci est suffisant pour traiter un petit nombre de données, mais lorsque nous allons vouloir traiter un plus gros volume de données, ceci peut devenir contraignant complexe à gérer.  
Par exemple, si nous voulons gérer un ensemble de points en géométrie, chacun comporte une abscisse et une ordonnée, il nous faudra donc deux valeurs par objet, si nous gérons un train, il sera composé d’une locomotive et d’un ou plusieurs wagons (dont nous ne connaissons pas le nombre à l’avance).  

Pour répondre à ce type de besoins, nous allons donc revoir trois nouveaux types d’objets : le type **tuple** (n-uplet), le type **list** (tableau indexé) et le type **dict** (dictionnaire).  

Ces types présentent des **points communs** mais aussi des **différences**, ils nous permettront d’adresser des besoins similaires et leur différences nous aideront à faire notre choix en fonction du problème à résoudre.


---
# p-uplets et p-uplets nommés (tuples)

## Définition

>📌 <font color="#FF0000">Un objet de type **tuple** est une suite ordonnée d’éléments qui peuvent être chacun de n’importe quel type. C’est une séquence d’éléments non modifiables.</font>  


## Création

Pour créer un tuple, on utilise les parenthèses et on sépare les éléments par des virgules :

````python
t = (3,7)
````

````python
t
````

````python
type(t)
````

Un tuple peut être vide

````python
t=()
````

````python
t
````

Ou contenir une seule valeur

````python
t=(4,)
````

````python
t
````

````python
type(t)
````

Attention à ne pas oublier la virgule, sinon le type sera différent

````python
t=(4)
````

````python
t
````

````python
type(t)
````

Il peut également contenir des objets de types différents

````python
t = (2,3,'a',8,)
````

````python
t
````

Un tuple peut contenir un autre tuple

````python
t = (2,3,'a',(6,7))
````

````python
t
````

La fonction ```tuple``` permet à partir d’une chaîne de caractères ou d’une liste (que nous verrons plus tard) de créer un tuple :

````python
t = tuple('Première NSI')
````

````python
t
````

## Opérations
- Concaténation

````python
t1 = (2,5)
t2 = (7,8)
t3 = t1 + t2
````

````python
t3
````

````python
t1 = t1 + (2,3)
````

````python
t1
````

- Multiplication

````python
t1 = (1,2)
t2 = 3*t1
````

````python
t2
````

## Accès aux éléments

Pour accéder aux éléments d’un tuple, on utilise les crochets :

````python
t = (2,5,6,3,0,3,6)
````

````python
t[2]
````

Comme pour les chaines de caractères, les indices des éléments commencent à 0 et se terminent à n-1 si n est le nombre d’éléments du tuple.  

On accèdera à la longueur d’un tuple avec la fonction ```len```

````python
len(t)
````

Un indice **négatif** permet de parcourir le tuple en partant de la fin (en commençant à -1)

````python
t[-2]
````

On peut accéder à un **tranche** de tuple en utilisant la syntaxe suivante :

````python
t[2:5]
````

````python
t[:5]
````

````python
t[3:]
````

````python
t[:3] + (10,) + t[4:]
````

Pour un tuple contenant un autre tuple, il faudra doubler les indices :

````python
t = ( ('a','b') , ('c','d') )
````

````python
t
````

````python
t[0]
````

````python
t[1][0]
````

>📌 <font color="#FF0000">Pour parcourir les éléments, on peut utiliser une boucle `for`, on dit que les tuples sont **itérables**</font>  

````python
t = (3,5,12)
for elt in t:
    print (elt)
````

Le tuple peut être également parcouru par ses indices :

````python
for i in range(len(t)):
    print (t[i])
````

Pour tester l’appartenance d’un élément à un tuple, on utilise l’opérateur ```in``` :

````python
t = ('a','b','c')
````

````python
'a' in t
````

````python
'd' in t
````

>📌 <font color="#FF0000">On ne peut pas modifier les éléments d’un tuple par affectation, on dit que les tuples sont **non mutables**</font>

````python
t[2] = 3
````

## Affectations multiples

La commande ```a,b = 1,2``` affecte 1 à la variable ```a``` et 2 à la variable ```b```. Ceci est en fait lié au fait que Python va affecter au tuple ```(a,b)``` la valeur du tuple ```(1,2)```.
Cette propriété est très utile pour permuter les valeurs de deux variables sans passer par une variable intermédiaire avec la commande ```a,b = b,a```

````python
a,b = 1,2

a,b = b,a
````

````python
a
````

````python
b
````

Comme pour les chaines de caractères, les indices des éléments commencent à 0 et se terminent à n-1 si n est le nombre d’éléments du tableau.  
 
On accèdera à la longueur d’un tableau avec la fonction ```len```

## Modification des tableaux
Pour ajouter des éléments, on peut utiliser la méthode ```append``` :

````python
l1 = [2,5]
l1.append(2)
l1.append(3)
````

````python
l1
````

Pour supprimer des éléments, on peut utiliser la fonction ```del``` ou méthode ```remove``` :

````python
l=[4, 1, 2, 6, 7, 15, 7]
del l[0]
````

````python
l
````

````python
l=[4, 1, 2, 6, 7, 15, 7]
l.remove(15)
````

````python
l
````

Pour insérer des éléments, on peut utiliser la méthode ```insert``` en lui passant en paramètres la position et l'élément à insérer :

````python
l=[4, 1, 2, 6, 7, 15, 7]
l.insert(3,99)
````

````python
l
````

Donc ```l.insert(0,x)``` permet d’ajouter x au début du tableau et ```l.insert(-1,x)``` permet d’ajouter x à la fin du tableau.

La méthode ```pop``` enlève du tableau l’élément situé à la position indiquée et retourne l’élément supprimé :

````python
l=[ 3 , 6 , 2 , 7 , 1 , 2 , 5]
l.pop(3)
````

````python
l
````

La méthode ```index``` retourne la position du premier élément du tableau ayant la valeur passée en paramètre :

````python
l=[ 3 , 6 , 2 , 7 , 1 , 2 , 5]
````

````python
l.index(7)
````

````python
l.index(2)
````

La méthode ```sort``` trie les éléments du tableau dans le tableau lui-même :

````python
l=[ 3 , 6 , 2 , 7 , 1 , 2 , 5]
````

````python
l.sort()
````

````python
l
````

La fonction ```sorted``` renvoie les éléments du tableau passé en paramètre dans un nouveau tableau :

````python
l=[ 3 , 6 , 2 , 7 , 1 , 2 , 5]
````

````python
sorted(l)
````

````python
l
````

La methode ```reverse``` inverse les éléments du tableau :

````python
l=[ 3 , 6 , 2 , 7 , 1 , 2 , 5]
````

````python
l.reverse()
````

````python
l
````

Copie d'un tableau :

````python
l1=[2,4,1,6,5,7]
l2=l1
l1.append(3)
````

````python
l1
````

````python
l2
````

> Vous noterez ici que lorsqu'on utilise l'opérateur `=` les tableaux l1 et l2 sont liés  
> Pour créer une copie d'un tableau, il faura utiliser la fonction ```list``` :

````python
l1=[2,4,1,6,5,7]
l2=list(l1)
l1.append(3)
````

Python permet également de générer un tableau en appliquant une fonction à un ensemble de
valeurs.  
Cela permet de créer un tableau en une seule ligne au lieu de passer par une boucle.  
>📌 <font color="#FF0000">Cette méthode est appelée **construction par compréhension**</font>  

````python
l=[x**2 for x in range(10)]
````

````python
l
````

Revient à écrire en une seule ligne le code suivant :

````python
l=[]
for x in range(10):
    l.append(x**2)
````

````python
l
````

Il est même possible de filtrer le tableau obtenu à l’aide d’une condition !  
On peut ainsi par exemple obtenir la liste des entiers inférieurs à 50 qui sont des multiples de 3 mais pas de 4 en une seule commande :

````python
l=[x for x in range(50) if x%3==0 and x%4!=0]
````

````python
l
````
